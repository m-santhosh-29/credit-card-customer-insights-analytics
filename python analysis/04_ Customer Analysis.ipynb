```python
# Import libraries for data analysis and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set a consistent style for charts
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
# Import pandas to work with CSV files
import pandas as pd

# Load the original customer dataset
customer = pd.read_csv(r"C:\Users\M.SANTHOSH\Downloads\Data Set\customer.csv")

# Load the additional customer dataset
cust_add = pd.read_csv(r"C:\Users\M.SANTHOSH\Downloads\Data Set\cust_add.csv")

# Combine both datasets into one customer DataFrame
customers = pd.concat([customer, cust_add], ignore_index=True)

# Display the total number of customer records
print("Total customer records:", customers.shape)
```

### Question 1: Which states have the highest number of customers?

**Why are we running this code?**

To identify the states with the largest customer base. This helps the business understand its geographical customer distribution and identify important markets.

```python
# Count customers in each state
state_counts = customers["state_cd"].value_counts().head(10)

# Display the top 10 states by customer count
print(state_counts)

# Create a bar chart to compare the top 10 states
state_counts.plot(kind="bar", rot=45)
plt.title("Question 1: Top 10 States by Customer Count")
plt.xlabel("State")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

### Question 2: Which age groups have the highest number of customers?

**Why are we running this code?**

To group customers by age and identify the age groups with the largest customer base. This helps the business understand which age groups it attracts most.

```python
# Group customer ages into age ranges
age_groups = pd.cut(
    customers["Customer_Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["Under 26", "26–35", "36–45", "46–55", "Above 55"]
)

# Count customers in each age group
age_counts = age_groups.value_counts().sort_index()

# Display the counts
print(age_counts)

# Create a bar chart
age_counts.plot(kind="bar", rot=0)
plt.title("Question 2: Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

### Question 3: What is the customer distribution by education level?



To compare the number of customers across education levels and understand the educational background of the customer base.

### Question 7: What is the average income by customer job category?

**Why are we running this code?**

To compare average income across job categories and identify which occupations have higher average customer incomes.

```python
# Calculate average income for each job category
average_income_by_job = customers.groupby("Customer_Job")["Income"].mean().sort_values(ascending=False)

# Display average income by job
print(average_income_by_job)

# Create a bar chart
average_income_by_job.plot(kind="bar", rot=45)
plt.title("Question 7: Average Income by Customer Job")
plt.xlabel("Customer Job")
plt.ylabel("Average Income")
plt.tight_layout()
plt.show()
```

### Question 8: What is the average customer satisfaction score by gender?

**Why are we running this code?**

To compare average customer satisfaction scores across gender categories and understand whether the recorded satisfaction scores differ between them.

```python
# Calculate average satisfaction score for each gender
satisfaction_by_gender = customers.groupby("Gender")["Cust_Satisfaction_Score"].mean()

# Display the averages
print(satisfaction_by_gender)

# Create a bar chart
satisfaction_by_gender.plot(kind="bar", rot=0)
plt.title("Question 8: Average Customer Satisfaction by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Satisfaction Score")
plt.tight_layout()
plt.show()
```

### Question 9: What is the average number of dependents by marital status?

**Why are we running this code?**

To compare the average number of dependents across marital status categories and understand differences in household characteristics.

```python
# Calculate average dependents by marital status
dependents_by_marital_status = customers.groupby("Marital_Status")["Dependent_Count"].mean()

# Display the averages
print(dependents_by_marital_status)

# Create a bar chart
dependents_by_marital_status.plot(kind="bar", rot=0)
plt.title("Question 9: Average Dependents by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Average Number of Dependents")
plt.tight_layout()
plt.show()
```

### Question 10: Which customer job categories have the highest average customer satisfaction score?

**Why are we running this code?**

To compare average satisfaction scores across customer job categories and identify which groups report higher or lower satisfaction scores.

```python
# Calculate average satisfaction score for each job category
satisfaction_by_job = customers.groupby("Customer_Job")["Cust_Satisfaction_Score"].mean().sort_values(ascending=False)

# Display the averages from highest to lowest
print(satisfaction_by_job)

# Create a bar chart
satisfaction_by_job.plot(kind="bar", rot=45)
plt.title("Question 10: Average Customer Satisfaction by Job")
plt.xlabel("Customer Job")
plt.ylabel("Average Satisfaction Score")
plt.tight_layout()
plt.show()
```
