### Question 8: What is the total transaction amount for each card category?

**Why are we running this code?**

To compare the total transaction amount across credit card categories and identify which category accounts for the highest transaction value.

```python
# Calculate total transaction amount for each card category
category_transactions = credit_cards.groupby("Card_Category")["Total_Trans_Amt"].sum()

# Display the totals
print(category_transactions)

# Create a bar chart to compare transaction amounts
category_transactions.plot(kind="bar", rot=0)
plt.title("Question 8: Total Transaction Amount by Card Category")
plt.xlabel("Card Category")
plt.ylabel("Total Transaction Amount")
plt.tight_layout()
plt.show()
```

### Question 9: Which expenditure types have the highest transaction amounts?

**Why are we running this code?**

To compare total transaction amounts across expenditure types and identify the categories where customers spend the most.

```python
# Calculate total transaction amount for each expenditure type
expense_transactions = credit_cards.groupby("Exp Type")["Total_Trans_Amt"].sum().sort_values(ascending=False)

# Display the totals from highest to lowest
print(expense_transactions)

# Create a bar chart to compare expenditure types
expense_transactions.plot(kind="bar", rot=45)
plt.title("Question 9: Total Transaction Amount by Expenditure Type")
plt.xlabel("Expenditure Type")
plt.ylabel("Total Transaction Amount")
plt.tight_layout()
plt.show()
```

### Question 10: What is the average credit limit for each credit card category?

**Why are we running this code?**

To compare the average credit limit across card categories and understand how credit limits differ between categories.

```python
# Calculate the average credit limit for each card category
average_credit_limit = credit_cards.groupby("Card_Category")["Credit_Limit"].mean().sort_values(ascending=False)

# Display the average credit limits
print(average_credit_limit)

# Create a bar chart to compare average credit limits
average_credit_limit.plot(kind="bar", rot=0)
plt.title("Question 10: Average Credit Limit by Card Category")
plt.xlabel("Card Category")
plt.ylabel("Average Credit Limit")
plt.tight_layout()
plt.show()
```
