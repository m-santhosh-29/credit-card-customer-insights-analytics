```python
# Import pandas to load and clean datasets
import pandas as pd

# Load the four original CSV files
customer = pd.read_csv(r"C:\Users\M.SANTHOSH\Downloads\Data Set\customer.csv")
cust_add = pd.read_csv(r"C:\Users\M.SANTHOSH\Downloads\Data Set\cust_add.csv")
credit_card = pd.read_csv(r"C:\Users\M.SANTHOSH\Downloads\Data Set\credit_card.csv")
cc_add = pd.read_csv(r"C:\Users\M.SANTHOSH\Downloads\Data Set\cc_add.csv")
```

```python
# Check the number of rows and columns in each dataset
for name, df in {
    "Customer": customer,
    "Additional Customer": cust_add,
    "Credit Card": credit_card,
    "Additional Credit Card": cc_add
}.items():
    print(f"{name}: {df.shape}")
```

```python
# Identify columns containing missing values
for name, df in {
    "Customer": customer,
    "Additional Customer": cust_add,
    "Credit Card": credit_card,
    "Additional Credit Card": cc_add
}.items():
    print(f"\n{name}:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Create a cleaned copy and remove columns that contain only missing values
cc_add_clean = cc_add.dropna(axis=1, how="all").copy()

# Verify the change
print("Original shape:", cc_add.shape)
print("Cleaned shape:", cc_add_clean.shape)
```

```python
# Count exact duplicate rows in each dataset
for name, df in {
    "Customer": customer,
    "Additional Customer": cust_add,
    "Credit Card": credit_card,
    "Additional Credit Card": cc_add_clean
}.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
# Convert transaction week dates from text to datetime format
credit_card["Week_Start_Date"] = pd.to_datetime(
    credit_card["Week_Start_Date"], errors="coerce"
)
cc_add_clean["Week_Start_Date"] = pd.to_datetime(
    cc_add_clean["Week_Start_Date"], errors="coerce"
)

# Check for dates that could not be converted
print("Credit Card invalid dates:", credit_card["Week_Start_Date"].isna().sum())
print("Additional Credit Card invalid dates:", cc_add_clean["Week_Start_Date"].isna().sum())
```

```python
# Rename the column to match the main dataset
cc_add_clean = cc_add_clean.rename(
    columns={"Total_Trans_Ct": "Total_Trans_Vol"}
)
```

```python
# Confirm that the customer datasets have matching column names
print("Customer columns match:", customer.columns.equals(cust_add.columns))

# Confirm that the credit-card datasets have matching column names
print("Credit-card columns match:", credit_card.columns.equals(cc_add_clean.columns))
```
